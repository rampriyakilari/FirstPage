```python
#Find Duplicate Number in Array
def find_duplicate(arr):
    seen = set()
    for number in arr:
        if number in seen:
            return number
        seen.add(number)
N = int(input())
arr = list(map(int, input().split()))
duplicate = find_duplicate(arr)
print(duplicate)
```

```python
#Count Set Bits
def count_set_bits(n):
    return bin(n).count('1')
T = int(input().strip())
res = []
for _ in range(T):
    N = int(input().strip())
    res.append(count_set_bits(N))
for res in res:
    print(res)
```
